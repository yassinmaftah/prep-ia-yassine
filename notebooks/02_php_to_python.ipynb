```python
tem = [10.5,5.5,12.99,15.3,22,5.8,44.5,18.5]
min = max = tem[0]
total = 0
for t in tem :
    if (t > max):
        max = t
    if (t < min):
        min = t
    total += t
avg = total / len(tem)
print(f"min : {min}, max : {max}, avg : {avg}")
```

```python
students = [
    {"nom": "Yassine", "note": 18, "filiere": "Dev Web"},
    {"nom": "Kamal", "note": 10, "filiere": "Marketing"},
    {"nom": "Aya", "note": 15, "filiere": "Data"}
]
```

```python
def good_students(students):
    return [s["nom"] for s in students if s["note"] >= 12]
print(good_students(students)) 
```

```python
for student in students:
    print("---")
    for key, value in student.items():
        print(f"{key}: {value}")
```

| **Tableaux associatifs** | `['clé' => 'valeur']` | `{'clé': 'valeur'}` (Dictionnaire) |
| **Concaténation** | Point `.` (`"Nom: " . $nom`) | f-string (`f"Nom: {nom}"`) |
| **Boucle foreach** | `foreach($liste as $item)` | `for item in liste:` |
| **Fonctions** | `function foo($x) { return $x; }`| `def foo(x): return x` |
| **Null/Vide** | `null` | `None` |
