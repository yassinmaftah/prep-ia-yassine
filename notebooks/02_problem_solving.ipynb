```python
def fu(numbers) :
    result = []
    for n in numbers:
        if (n % 2 == 0):
            result.append(n*10)
    return result
print(fu([1,2,3,4,5,6]))
```

```python
def fu(numbers) :
    return [n * 10 for n in numbers if n % 2 == 0]
print(fu([1,2,3,4,5,6]))
```

```python
def Count_Letters(text) :
    counts = {}
    for letter in text:
        if letter not in counts:
            counts[letter] = 0
        counts[letter]+=1
    return counts
print(Count_Letters("yasyyyyyyyyyyyyyyysine"))
```

```python
def get_adult_names(users_list):
    return [u['name'] for u in users_list if u['age'] >= 18 ]
users = [
    {"name": "Yassine", "age": 21},
    {"name": "Aya", "age": 16},
    {"name": "Kamal", "age": 30},
    {"name": "Rania", "age": 17}
]

print(get_adult_names(users))
```

```python
def sum_by_category(sales):
    totals = {}
    for s in sales:
        totals[s["category"]] = totals.get(s["category"],0) + s["amount"]
    
    return totals

orders = [
    {"category": "Electronics", "amount": 120},
    {"category": "Clothing", "amount": 40},
    {"category": "Electronics", "amount": 250},
    {"category": "Food", "amount": 15},
    {"category": "Clothing", "amount": 60}
]

# This should print: {'Electronics': 370, 'Clothing': 100, 'Food': 15}
print(sum_by_category(orders))
```

```python
class User:
    def __init__(self,name, age):
        self.name = name
        self.age = age
    def is_adult(self):
        return self.age > 17

user1 = User("Yassine", 21)
user2 = User("Aya", 16)

print(f"{user1.name} est adulte : {user1.is_adult()}") 

print(f"{user2.name} est adulte : {user2.is_adult()}")
```

```python
def find_two_sum(numbers, target):
    seen = []
    for n in range(0, len(numbers)):
        for j in range(n + 1,len(numbers)):
            if (numbers[n] + numbers[j] == target):
              seen.append([n,j])  
        
    return seen
my_numbers = [2, 7, 11, 15,5]
my_target = 9
print(find_two_sum(my_numbers, my_target))
```
