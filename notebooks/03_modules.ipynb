### The difference between running `main.py` and importing in a notebook:

When I run `python main.py` in the terminal, Python executes the file **directly**. In this case, Python's special built-in variable `__name__` is assigned the value `"__main__"`. The script knows it is the main program being run, so all the code inside the `if __name__ == "__main__":` block activates and executes the `print()` statements.

Conversely, when I use `import` in this notebook, I am telling Python: *"Just fetch the functions from the toolbox, but do not run the program."* Because the file is only being imported and not executed directly, the `if __name__ == "__main__":` block is completely ignored. This is why no text printed out automatically when we imported the module!
